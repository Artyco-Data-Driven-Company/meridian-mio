```python
# Definir constantes
PKL_MODEL = "model.pkl"
PRODUCT = "auto"
START_DATE = "2025-01-01"
END_DATE = "2025-12-01"

UPPER_BUDGET = 6489.84
LOWER_RULES_U = {
}
UPPER_RULES_U = {
    "medios-digital-performance": 0,
    "medios-digital-branding": 0
}

UPPER_BUDGET = 6189.84
LOWER_RULES_U_R = {
}
UPPER_RULES_U_R = {
    "medios-digital-performance": 0.07,
    "medios-digital-branding": 0.07,
    "medios-patrocinio": 0.07
}

LOWERER_BUDGET = 4489.84
LOWER_RULES_L = {
    "medios-patrocinio": 0,
    "medios-off": 0
}
UPPER_RULES_L = {
}

PCT_SPEND = [0.01638, 0.07, 0.7733, 0.1289, 0.01142]

SPEND_RULES_HARDCODED = {}
```

```python
from meridian.analysis import optimizer, ClientConfig
import joblib
import pandas as pd

mmm = joblib.load(PKL_MODEL)
```

```python
mmm.input_data.get_all_paid_channels()
```

```python
# Crear configuración del cliente
client_config = ClientConfig(config_path="sandbox\\new_optimization\\config.yaml")
```

```python
channels = mmm.input_data.get_all_paid_channels()
lower_constraints = []
upper_constraints = []

for ch in channels:
  if ch in LOWER_RULES_U:
    lower_constraints.append(LOWER_RULES_U[ch])
  else:
    lower_constraints.append(0.5)
  if ch in UPPER_RULES_U:
    upper_constraints.append(UPPER_RULES_U[ch])
  else:
    upper_constraints.append(0.5)
```

```python
# ========================================
# OPTIMIZACIÓN PRESUPUESTARIA 0
# ========================================
budget_optimizer = optimizer.BudgetOptimizer(mmm, client_config)

optimization_results = budget_optimizer.optimize(
    use_posterior=True,
    selected_times=None,
    fixed_budget=True,
    budget=None,
    start_date=START_DATE,
    end_date=END_DATE,
    spend_constraint_lower=lower_constraints,
    spend_constraint_upper=upper_constraints,
    target_roi=None,
    target_mroi=None,
    gtol=0.0001,
    use_optimal_frequency=True,
    use_kpi=True,
    confidence_level=0.9,
    batch_size=3000
)

optimization_results.output_optimization_summary(
    filename="optimization_output_0.html",
    filepath="sandbox\\new_optimization\\",
)
```
